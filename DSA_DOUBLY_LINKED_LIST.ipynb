```python
from datetime import datetime

class Node:
    def __init__(self, data):
        self.data = data  # [description, priority, category, due date, recurrence]
        self.prev = None
        self.next = None

    def __str__(self):
        return (f"Task: {self.data[0]}, Priority: {self.data[1]}, Category: {self.data[2]}, "
                f"Due: {self.data[3].strftime('%d-%m-%Y')}, Recurrence: {self.data[4]}")

class DoublyLinkedList:
    def __init__(self):
        self.head = None
        self.tail = None
    
    def add_task(self, task):
        if not self.head:
            self.head = self.tail = task
        else:
            self.tail.next = task
            task.prev = self.tail
            self.tail = task
        print(f"Task '{task.data[0]}' added successfully!")
        
    def remove_task(self, description):
        temp = self.head
        while temp:
            if temp.data[0] == description:  # Check description
                if temp.prev:
                    temp.prev.next = temp.next
                if temp.next:
                    temp.next.prev = temp.prev
                if temp == self.head:
                    self.head = temp.next
                if temp == self.tail:
                    self.tail = temp.prev
                print(f"Task '{description}' completed and removed.")
                return True
            temp = temp.next
        print("Task not found!")
        return False

    def display_tasks(self):
        temp = self.head
        if not temp:
            print("No tasks available.")
            return
        print("\nPending Tasks:")
        while temp:
            print(temp)
            temp = temp.next
            
    
    def reverse_display(self):
        temp = self.tail
        if not temp:
            print("No tasks available.")
            return
        print("\nTasks in Reverse Order:")
        while temp:
            print(temp)
            temp = temp.prev

    def count_tasks(self):
        count = 0
        temp = self.head
        while temp:
            count += 1
            temp = temp.next
        return count

    def clear_tasks(self):
        self.head = self.tail = None
        print("All tasks have been cleared!")

def main():
    task_list = DoublyLinkedList()

    while True:
        print("\nTASK MANAGER MENU")
        print("1. Add Task")
        print("2. Complete Task")
        print("3. Display Tasks")
        print("4. Reverse Display")
        print("5. Count Tasks")
        print("6. Clear Tasks")
        print("7. Exit")
        choice = input("Enter choice: ")

        if choice == '1':
            description = input("Enter task description: ")
            priority = int(input("Enter task priority: "))
            category = input("Enter category: ")
            due_date = input("Enter due date (dd-MM-yyyy): ")

            try:
                due_date_obj = datetime.strptime(due_date, '%d-%m-%Y')  # Convert string to datetime
            except ValueError:
                print("Invalid date format! Please enter in dd-MM-yyyy format.")
                continue

            recurrence = input("Enter recurrence: ")
            new_task = Node([description, priority, category, due_date_obj, recurrence])
            task_list.add_task(new_task)

        elif choice == '2':
            desc = input("Enter the task to complete: ")
            task_list.remove_task(desc)

        elif choice == '3':
            task_list.display_tasks()

        elif choice == '4':
            task_list.reverse_display()

        elif choice == '5':
            print(f"Total pending tasks: {task_list.count_tasks()}")

        elif choice == '6':
            task_list.clear_tasks()

        elif choice == '7':
            print("Goodbye!")
            break

        else:
            print("Invalid choice! Try again.")

if __name__ == "__main__":
    main()
```
