### **<span style="background-color:lightblue;"> METHODOLOGY </span>**
<div style="text-align: justify; font-size: 18px;"><br>
<span style="background-color:lightpink;"> TRIAL 1: </span>

spaCy: is a powerful open-source Natural language processing (NLP) library designed for text processing. One of its key features, Named Entity Recognition (NER), helps to identify important/spefic entities within text. spaCy’s pre-trained deep learning models make NER highly accurate and adaptable for real-world applications.

In this project, spaCy is used to parse the input text/command from user to extract locations, get event start/end time and dates, ensuring that events are correctly recognised for appropriate time blocking in the Google Calender. Combining spaCy's linguistic processing with "Personal assistant" specific rules improves the accuracy of event scheduling by ensuring that the details are captured correctly.

Moreover, DateFinder is another Python library designed to extract dates and times from natural language text. It uses pattern matching to detect various date formats, such as "January 25th, 2025", "tomorrow at 6 PM", or "Monday at noon", and converts them into datetime objects.

By combining DateFinder with spaCy's NER, we aim to effectively handle both explicit (e.g., "February 10th") and relative (e.g., "tomorrow") date expressions, making the event scheduling process seamless. However, this approach does not yield satisfactory results across all test cases. In particular, extracting meeting locations remains inefficient, necessitating a more advanced NLP-based extraction method for improved accuracy and suitability across various test cases.
    
<span style="background-color:lightpink;"> TRIAL 2: </span>

This method allows extracting structured event information from natural language input using Google Gemini AI and NLP techniques. The script takes a user-provided text description of an event and extracts key details such as the event summary, date, time, and location. This is also appropriately manipulated to handle relative dates (e.g., "tomorrow" or "next Monday"), recognizing time ranges and implicit time references (e.g., "morning" → "08:00-11:59"), and normalizing locations with the Geopy library.
    
This method involves the use of Google Gemini AI to generate a JSON output containing the structured event details. To ensure the output is valid, a regex-based function extracts JSON data from the AI's response. This also includes error handling with retries to prevent failures due to API issues. If the AI fails to generate a response, a default fallback ensures that the event is still scheduled, using the current date and marking missing information appropriately.
    
For date extraction, in this method a regex is employed and the dateparser library is used to handle multiple formats, including specific dates, date ranges, and relative time expressions like "day after tomorrow." If no date is found in the input, it defaults to the current date. Similarly, the time details are extracted by identifying explicit times (e.g., "10:00 AM"), time ranges (e.g., "9 AM to 12 PM"), and time-related keywords such as "afternoon" or "night," mapping them to predefined time slots.
    
Location extraction is performed by splitting location mentions from the input text and normalizing them using Geopy. If Geopy cannot verify a location, the script retains the original input. This ensures that even ambiguous or less-known locations are still included. The program runs in a loop, allowing users to continuously enter event details or exit by typing "exit."

### **<span style="background-color:lightblue;"> CONCLUSION </span>**
<div style="text-align: justify; font-size: 18px;"><br>
Our project model currently processes both text and voice input (using Google Text-to-Speech) to extract key details such as start time, event date, event location, event description, and schedule the event in Google Calendar using time blocking. 

Moving forward, our primary focus will be on enhancing the extraction techniques by testing the model across a wide range of text inputs and real-world scenarios. Additionally, we aim to improve its ability to accept event durations . This modification will allow users to specify the exact duration of their events for more accurate scheduling.  

Furthermore, we plan to make the system more interactive by introducing a feature where it can intelligently suggest suitable time slots based on the user's calendar availability. For instance, if a user states, “I need to finish this task by February 13th, and it will take about two hours,” the system should analyze their schedule and propose an optimal time. A possible response could be:  

*"Considering your free time on February 7th, shall I schedule this task from 3:00 PM to 5:00 PM on Saturday?"*  

By integrating such a feature, our system will go beyond basic scheduling and evolve into a smart assistant capable of proactive time management. These enhancements will make event scheduling more efficient, flexible, and user-friendly.
